```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/Mahroosa/Girlstocodedatabase2026/main/maldives_feedback.csv")
sns.barplot(data=df, x='Location' , y='Rating')
sns.set_theme(style="whitegrid")

plt.title('Average rating for each Industry based on Islands')
plt.xlabel('Industry Category')
plt.ylabel('Rating')
plt.show()
```
